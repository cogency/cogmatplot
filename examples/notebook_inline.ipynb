# Inline figures

Static output in the notebook: an image with an overlay, a 2D plot, a 3D line, and a point cloud (xyz+rgb plus a solid-color tag). These figures are not interactive. Use `cogmatplot.use("qt")` when you want pop-out windows on another monitor.


```python
%matplotlib inline
import sys
from pathlib import Path

here = Path.cwd()
if (here / "scene.py").exists():
    sys.path.insert(0, str(here))
elif (here / "examples" / "scene.py").exists():
    sys.path.insert(0, str(here / "examples"))

import cogmatplot
cogmatplot.use("inline")
from cogmatplot import FigureManager, closeAll
from scene import draw, make
```

```python
closeAll()
fm = FigureManager(2, 2, pause=False)
draw(fm, n=800)
```

```python
live = FigureManager(1, 1, pause=False)
live.imshow(0, 0, make(n=10)["image"], viewer="cv", title="inline image")
live.drawnow()
```
